```python
# Step 2.1: Import the necessary libraries
import requests
from bs4 import BeautifulSoup
```

Get Title & Price:


```python
# Step 2.2: Define the URL you want to scrape
url = 'https://www.zameen.com/Homes/Karachi_Bahria_Town_Karachi_Bahria_Town___Precinct_11-10019-1.html'


# Step 2.3: Send a GET request to the URL
response = requests.get(url)


# Step 2.4: Check if the request was successful
if response.status_code == 200:
    # Step 2.5: Parse the page content with BeautifulSoup
    soup = BeautifulSoup(response.content, 'html.parser')


    # Step 2.6: Extract and print the page title
    title = soup.title.string
    print('Page Title:', title)


    # Step 2.7: Find property listings on the page
    listings = soup.find_all('li', class_='a37d52f0')


    # Step 2.8: Loop through each listing and extract information
    for listing in listings:
        # Extract the title and price of each listing
        listing_title = listing.find('h2', class_='_36dfb99f').text.strip() if listing.find('h2', class_='_36dfb99f') else 'N/A'
        listing_price = listing.find('span', class_='dc381b54').text.strip() if listing.find('span', class_='dc381b54') else 'N/A'
       
        # Print the extracted information
        print('Title:', listing_title)
        print('Price:', listing_price)
        print('---')


else:
    # Step 2.9: Handle request errors
    print('Failed to retrieve the page. Status code:', response.status_code)
```

** CODE WITH PAGINATION;

```python
# Step 1: Define the base URL
base_url = 'https://www.zameen.com/Homes/Karachi_Bahria_Town_Karachi_Bahria_Town___Precinct_11-10019-1.html'


# Step 2: Initialize the page counter
page = 1


while True:
    # Step 3: Create the URL for the current page
    url = f"{base_url}?page={page}"
    response = requests.get(url)


    # Step 4: Check if the request was successful
    if response.status_code == 200:
        soup = BeautifulSoup(response.content, 'html.parser')
        title = soup.title.string
        print('Page Title:', title)


        # Step 5: Find property listings
        listings = soup.find_all('li', class_='a37d52f0')
        # Check if there are no listings
        if not listings:
            print("No more listings found.")
            break


        # Step 6: Loop through each listing and extract information
        for listing in listings:
            # Extract property title
            listing_title = listing.find('h2', class_='_36dfb99f').text.strip() if listing.find('h2', class_='_36dfb99f') else 'N/A'
           
            # Extract price
            listing_price = listing.find('span', class_='dc381b54').text.strip() if listing.find('span', class_='dc381b54') else 'N/A'
           
            # Extract location
            location = listing.find('div', class_='db1aca2f').text.strip() if listing.find('div', class_='db1aca2f') else 'N/A'
           
            # Extract property type
            property_type = listing.find('span', class_='_1778c16d').text.strip() if listing.find('span', class_='_1778c16d') else 'N/A'
           
            # Extract size
            size = listing.find('div', class_='_52908f50').text.strip() if listing.find('div', class_='_52908f50') else 'N/A'
           
            # Extract number of bedrooms
            bedrooms = listing.find_all('span', class_='_6d9b9b83')
            bedrooms_count = bedrooms[0].text.strip() if len(bedrooms) > 0 else 'N/A'
           
            # Extract number of bathrooms
            bathrooms_count = bedrooms[1].text.strip() if len(bedrooms) > 1 else 'N/A'
           
            # Extract description
            description = listing.find('span', class_='_74f91f40').text.strip() if listing.find('span', class_='_74f91f40') else 'N/A'


            # Print extracted information
            print('Title:', listing_title)
            print('Price:', listing_price)
            print('Location:', location)
            print('Property Type:', property_type)
            print('Size:', size)
            print('Bedrooms:', bedrooms_count)
            print('Bathrooms:', bathrooms_count)
            print('Description:', description)
            print('---')


        # Step 7: Increment the page counter
        page += 1
    else:
        print('Failed to retrieve the page. Status code:', response.status_code)
        break
```
